# variable length argument 

```python
def person(name, age):
    print(name)
    print(age)
    
person('nit', 40) 
```

```python
def person(name, age):
    print(name)
    print(age)
    
person('nit', 40, 50, 60, 70) 
```

```python
def person(name, *age):   # in variable length the last argument is written with  * (star)
    print(name)
    print(age)
    
person('nit', 40, 50, 60, 70) 
```

```python
def person(name, *age):
    print(type(name))
    print(type(age))
    
person('nit', 40, 50, 60, 70) 
```

# local variable vs global variable

Global Variable
      ↓

Inside Function  ✔

Outside Function ✔
✅ Global variables can be accessed both inside and outside functions.

------------------------------------

Local Variable
      ↓

Inside Function  ✔

Outside Function ✖

A global variable is declared outside a function and can be accessed throughout the program. A local variable is declared inside a function and can only be accessed within that function.

```python
a = 10  #-- globla variable 

def something():
    
    b = 15 #local variable
    
    print('in function',b)    
    print('out function',a)  
```

```python
a = 10  #-- globla variable 

def something():
    
    b = 15 #local variable
    
    print('in function',b)   
    
print('out function',a)  
```

```python
# Example 1: Access Global Variable Inside Function

a = 10   # Global Variable

def test():
    print("Inside Function:", a)

test()

print("Outside Function:", a)
```

```python
# Example 2: Access Local Variable

def test():
    b = 20   # Local Variable
    print("Inside Function:", b)

test()

print("Outside Function:", b)
```

```python
a = 10   # Global variable

class nit:
    c = 15   # Class variable

def something():
    b = 20   # Local variable
    print("Inside function, b =", b)
    print("Inside function, c =", nit.c)

something()

print("Outside function, a =", a)
```
